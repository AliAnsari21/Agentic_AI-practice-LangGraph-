```python
from langgraph.store.memory import InMemoryStore
```

```python
#creating a store
store=InMemoryStore()
```

```python
#creating a namespace
namespace=("user","u1")
```

```python
#adding memories
store.put(namespace,"1",{"data":"user likes pizza"})
store.put(namespace,"2",{"data":"user prefers dark mode"})
```

```python
#another namespace
namespace2=("user","u2")

#add memories
store.put(namespace2,"1",{"data":"user likes pasta"})
store.put(namespace2,"2",{"data":"user prefers grid style navigation"})
```

Retrieving Memories

```python
store.get(namespace,"1")
```

Retrieving all memories

```python
items=store.search(namespace2)
for item in items:
    print(item.value)
```

Semantic Search

```python
from langchain_huggingface import HuggingFaceEmbeddings
from dotenv import load_dotenv
load_dotenv()
embedding_model = HuggingFaceEmbeddings(
    model_name="Qwen/Qwen3-Embedding-0.6B"
)
```

```python
store=InMemoryStore(index={'embed':embedding_model,'dims':1536})
```

```python
namespace=("user","u1")
```

```python
store.put(namespace, "1", {"data": "user likes pizza"})
store.put(namespace, "2", {"data": "user likes cricket"})
store.put(namespace, "3", {"data": "user likes Python programming"})
store.put(namespace, "4", {"data": "user is learning data science"})
store.put(namespace, "5", {"data": "user likes artificial intelligence"})
store.put(namespace, "6", {"data": "user prefers watching movies"})
store.put(namespace, "7", {"data": "user is interested in machine learning"})
store.put(namespace, "8", {"data": "user likes travelling"})
store.put(namespace, "9", {"data": "user is learning generative AI"})
store.put(namespace, "10", {"data": "user likes listening to music"})
```

```python
items=store.search(namespace,query="what is user currently learning",limit=1)
for item in items:
    print(item.value)
```

```python
items=store.search(namespace,query="what are user preferences",limit=3)
for item in items:
    print(item.value)
```
